```python
import os
import shutil
import subprocess
import sys

# Kaggle input locations
MODEL_ROOT = "/kaggle/input/datasets/suryadhakshinesh/modelfiles/New folder"
DATA_PATH = "/kaggle/input/datasets/suryadhakshinesh/healthcare-fraud-detection/provider_data_synthetic_15000_40yes_60no.csv"

# Create a clean working project
WORK = "/kaggle/working/fraud-api"

if os.path.exists(WORK):
    shutil.rmtree(WORK)

os.makedirs(WORK)
os.makedirs(os.path.join(WORK, "data"))
os.makedirs(os.path.join(WORK, "models"))
os.makedirs(os.path.join(WORK, "src"))
os.makedirs(os.path.join(WORK, "training"))

# Copy model
shutil.copytree(
    os.path.join(MODEL_ROOT, "models"),
    os.path.join(WORK, "models"),
    dirs_exist_ok=True
)

# Copy source code
shutil.copytree(
    os.path.join(MODEL_ROOT, "src"),
    os.path.join(WORK, "src"),
    dirs_exist_ok=True
)

# Copy retraining script
shutil.copy(
    os.path.join(MODEL_ROOT, "training", "retrain.py"),
    os.path.join(WORK, "training", "retrain.py")
)

# Copy training CSV to the filename expected by the pipeline
shutil.copy(
    DATA_PATH,
    os.path.join(WORK, "data", "training_snapshot.csv")
)

print("Project prepared at:", WORK)
print("\nFiles:")
for root, dirs, files in os.walk(WORK):
    for file in files:
        print(os.path.join(root, file))

# Run the existing retraining pipeline
print("\n" + "="*70)
print("STARTING V2 RETRAINING")
print("="*70 + "\n")

result = subprocess.run(
    [sys.executable, "training/retrain.py"],
    cwd=WORK,
    capture_output=True,
    text=True
)

print(result.stdout)

if result.stderr:
    print("\n--- STDERR ---")
    print(result.stderr)

print("\nExit code:", result.returncode)
```

```python
import os
import json
import shutil

WORK = "/kaggle/working/fraud-api"
OUTPUT = "/kaggle/working/final_model"

# Clean previous output
if os.path.exists(OUTPUT):
    shutil.rmtree(OUTPUT)

os.makedirs(OUTPUT)

# Check which model is active
active_model_file = os.path.join(WORK, "active_model.json")

if os.path.exists(active_model_file):
    with open(active_model_file, "r") as f:
        active = json.load(f)
else:
    active = {"version": "v1"}

version = active["version"]

print("Active model:", version)

# Copy the active model
source_model = os.path.join(WORK, "models", version)
destination_model = os.path.join(OUTPUT, "models", version)

shutil.copytree(source_model, destination_model)

# Copy active model pointer
shutil.copy(
    active_model_file,
    os.path.join(OUTPUT, "active_model.json")
)

# Create a simple manifest
manifest = {
    "active_model": version,
    "source_dataset": "provider_data_synthetic_15000_40yes_60no.csv",
    "features": 30
}

with open(os.path.join(OUTPUT, "manifest.json"), "w") as f:
    json.dump(manifest, f, indent=2)

print("\nFINAL MODEL ARTIFACT")
for root, dirs, files in os.walk(OUTPUT):
    for file in files:
        print(os.path.join(root, file))
```
