```python
target_df = pd.read_csv(
    "/kaggle/input/datasets/divyanths/provider/Train-1542865627584.csv"
)

print("Target shape:", target_df.shape)
print("Target columns:", target_df.columns.tolist())
```

```python
print("========== BENEFICIARY ==========")
print("Unique BeneID:", beneficiary_df["BeneID"].nunique())


print("\n========== INPATIENT ==========")
print("Unique BeneID:", inpatient_df["BeneID"].nunique())
print("Unique ClaimID:", inpatient_df["ClaimID"].nunique())
print("Unique Provider:", inpatient_df["Provider"].nunique())


print("\n========== OUTPATIENT ==========")
print("Unique BeneID:", outpatient_df["BeneID"].nunique())
print("Unique ClaimID:", outpatient_df["ClaimID"].nunique())
print("Unique Provider:", outpatient_df["Provider"].nunique())


print("\n========== TARGET ==========")
print("Unique Provider:", target_df["Provider"].nunique())
print("Target rows:", len(target_df))
```

```python
target_providers = set(target_df["Provider"])
inpatient_providers = set(inpatient_df["Provider"])
outpatient_providers = set(outpatient_df["Provider"])

print("Target providers:", len(target_providers))

print(
    "Target providers present in Inpatient:",
    len(target_providers & inpatient_providers)
)

print(
    "Target providers present in Outpatient:",
    len(target_providers & outpatient_providers)
)

print(
    "Target providers present in BOTH:",
    len(
        target_providers
        & inpatient_providers
        & outpatient_providers
    )
)

print(
    "Target providers only in Inpatient:",
    len(
        (target_providers & inpatient_providers)
        - outpatient_providers
    )
)

print(
    "Target providers only in Outpatient:",
    len(
        (target_providers & outpatient_providers)
        - inpatient_providers
    )
)

print(
    "Target providers in neither:",
    len(
        target_providers
        - inpatient_providers
        - outpatient_providers
    )
)
```

```python
beneficiary_ids = set(beneficiary_df["BeneID"])
inpatient_benes = set(inpatient_df["BeneID"])
outpatient_benes = set(outpatient_df["BeneID"])

print("========== BENEFICIARY MAPPING ==========")

print(
    "Beneficiaries in Beneficiary dataset:",
    len(beneficiary_ids)
)

print(
    "Inpatient beneficiaries:",
    len(inpatient_benes)
)

print(
    "Outpatient beneficiaries:",
    len(outpatient_benes)
)

print(
    "\nInpatient BeneIDs present in Beneficiary:",
    len(inpatient_benes & beneficiary_ids)
)

print(
    "Inpatient BeneIDs NOT in Beneficiary:",
    len(inpatient_benes - beneficiary_ids)
)

print(
    "\nOutpatient BeneIDs present in Beneficiary:",
    len(outpatient_benes & beneficiary_ids)
)

print(
    "Outpatient BeneIDs NOT in Beneficiary:",
    len(outpatient_benes - beneficiary_ids)
)

print(
    "\nBeneficiaries appearing in BOTH Inpatient and Outpatient:",
    len(inpatient_benes & outpatient_benes)
)
```

```python
print("========== PROVIDER-BENEFICIARY RELATIONSHIP ==========")

inpatient_provider_benes = (
    inpatient_df.groupby("Provider")["BeneID"]
    .nunique()
)

outpatient_provider_benes = (
    outpatient_df.groupby("Provider")["BeneID"]
    .nunique()
)

print("\nInpatient:")
print(inpatient_provider_benes.describe())

print("\nOutpatient:")
print(outpatient_provider_benes.describe())
```

```python
print("\nTotal unique provider-beneficiary pairs:")

print(
    "Inpatient:",
    inpatient_df[["Provider", "BeneID"]].drop_duplicates().shape[0]
)

print(
    "Outpatient:",
    outpatient_df[["Provider", "BeneID"]].drop_duplicates().shape[0]
)
```

```python
print("========== PROVIDER-CLAIM RELATIONSHIP ==========")

inpatient_claims_per_provider = (
    inpatient_df.groupby("Provider")["ClaimID"]
    .nunique()
)

outpatient_claims_per_provider = (
    outpatient_df.groupby("Provider")["ClaimID"]
    .nunique()
)

print("\nInpatient claims per provider:")
print(inpatient_claims_per_provider.describe())

print("\nOutpatient claims per provider:")
print(outpatient_claims_per_provider.describe())
```

```python
print("\nMissing Provider values:")

print(
    "Inpatient:",
    inpatient_df["Provider"].isna().sum()
)

print(
    "Outpatient:",
    outpatient_df["Provider"].isna().sum()
)
```

```python
print("\nMissing ClaimID values:")

print(
    "Inpatient:",
    inpatient_df["ClaimID"].isna().sum()
)

print(
    "Outpatient:",
    outpatient_df["ClaimID"].isna().sum()
)
```

```python
print("========== TARGET DISTRIBUTION ==========")

print(target_df["PotentialFraud"].value_counts())

print("\nPercentage distribution:")

print(
    target_df["PotentialFraud"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
inpatient_claims = set(inpatient_df["ClaimID"])
outpatient_claims = set(outpatient_df["ClaimID"])

print("Inpatient ClaimIDs:", len(inpatient_claims))
print("Outpatient ClaimIDs:", len(outpatient_claims))

print(
    "ClaimIDs appearing in BOTH:",
    len(inpatient_claims & outpatient_claims)
)
```

```python
inpatient_pairs = set(
    zip(inpatient_df["Provider"], inpatient_df["BeneID"])
)

outpatient_pairs = set(
    zip(outpatient_df["Provider"], outpatient_df["BeneID"])
)

print("Inpatient Provider-BeneID pairs:", len(inpatient_pairs))
print("Outpatient Provider-BeneID pairs:", len(outpatient_pairs))

print(
    "Provider-BeneID pairs appearing in BOTH:",
    len(inpatient_pairs & outpatient_pairs)
)
```
